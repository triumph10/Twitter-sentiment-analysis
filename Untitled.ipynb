### Data Cleaning
1. Removing white space
2. Removing not numbers or letter characters
4. Lowercasing
5. Removeing stop words
6. Removing URLs
7. Removign Number
8. Removing specical character

```python
def text_cleaning(text, stopwords):
    # Remove extra spaces between words
    text = text.str.replace(r'\s+', ' ', regex=True)
    
    # Remove any non-alphabetic characters (no numbers, special characters)
    text = text.apply(lambda x: re.sub(r'[^a-zA-Z\s]', '', x))
    
    # Convert all characters to lowercase for consistency (because 'Happy' and 'happy' are the same)
    text = text.str.lower()
    
    # Remove stopwords like 'the', 'and', 'is' since they don't contribute to the mood
    text = text.apply(lambda x: ' '.join([word for word in x.split() if word not in stopwords]))
    
    # Remove numbers (text like 'I feel 100%' isn't helpful for mood detection)
    text = text.str.replace(r'\d+', '', regex=True)
    
    # Remove any URLs that might appear in text
    text = text.str.replace(r'http\S+', '', regex=True)
    
    # Remove special characters like punctuation (so '!wow' becomes 'wow')
    text = text.str.replace(r'[^\w\s]', '', regex=True)
    
    return text
```
